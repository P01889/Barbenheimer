A matrix can transform vectors by stretching, shrinking, rotating, or reflecting them. An eigenvector is a special vector whose direction does not change under the transformation. When the matrix is applied to an eigenvector, the vector is only scaled by a factor called the eigenvalue.

- If |λ| > 1, the vector is stretched.
- If 0 < |λ| < 1, the vector is shrunk.
- If λ < 0, the vector is also reversed in direction.
- If λ = 1, the vector remains unchanged.

Geometrically, stretching a vector means changing its length while keeping it on the same line (same or opposite direction).

### Geometric Meaning of Eigenvalues and Eigenvectors

When a matrix transforms space, most vectors change both their **length** and **direction**. However, an **eigenvector** is a special vector whose direction remains unchanged after the transformation. The matrix only scales it by a number called the **eigenvalue**.

For an eigenvector **v** and eigenvalue **λ**:

A v = λ v

Geometrically:

- If **λ > 1**, the vector is **stretched** (made longer).
- If **0 < λ < 1**, the vector is **squished** or **compressed** (made shorter).
- If **λ < 0**, the vector is scaled and also **reversed in direction**.
- If **λ = 1**, the vector is unchanged.

You can think of a matrix as a transformation that bends, rotates, stretches, or compresses space. Eigenvectors identify the special directions that are not rotated by the transformation, while eigenvalues tell us how much stretching or compression occurs along those directions.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.

4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = None, None, None
# YOUR CODE HERE
U, S, Vt = np.linalg.svd(X, full_matrices=False)

X_reconstructed = None
# YOUR CODE HERE — reconstruct from U, S, Vt
X_reconstructed = U @ np.diag(S) @ Vt

X_approx = None
# YOUR CODE HERE — rank-1 approximation
X_approx = S[0] * np.outer(U[:, 0], Vt[0, :])

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

### SVD → PCA Connection

In Singular Value Decomposition (SVD), a matrix X is decomposed as:

X = UΣVᵀ

where:

- **U** contains the left singular vectors. These represent directions in the original observation space.
- **Σ (S)** contains the singular values. They measure the importance or strength of each component.
- **Vᵀ** contains the right singular vectors. These represent the principal directions (axes) in feature space.

For PCA, after centering the data:

- The **columns of V** (or rows of Vᵀ) are the **principal components**.
- The **singular values** indicate how much variance is captured by each principal component.
- Larger singular values correspond to directions with greater variance in the data.

Therefore, the matrix that contains the **directions of maximum variance** is **V** (equivalently, the rows of Vᵀ).

PCA can be performed using SVD by keeping only the largest singular values and their corresponding singular vectors, which reduces dimensionality while preserving most of the data's variance.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = None
# YOUR CODE HERE
ks_s1 = stats.kstest(s1, 'norm')

ks_s2 = None
# YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')

print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = None
# YOUR CODE HERE — adfuller(ts)
adf_result = adfuller(ts)

print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = None
# YOUR CODE HERE — np.diff
ts_diff = np.diff(ts)

adf_diff = None
# YOUR CODE HERE
adf_diff = adfuller(ts_diff)

print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

3. In the markdown cell, define: *concept drift*, *covariate drift*, PSI thresholds (what values indicate no/minor/major shift), and one trigger for retraining.

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    epsilon = 1e-10

    breakpoints = np.histogram_bin_edges(expected, bins=bins)

    expected_counts, _ = np.histogram(expected, bins=breakpoints)
    actual_counts, _ = np.histogram(actual, bins=breakpoints)

    expected_pct = expected_counts / len(expected)
    actual_pct = actual_counts / len(actual)

    psi = np.sum(
        (actual_pct - expected_pct) *
        np.log((actual_pct + epsilon) / (expected_pct + epsilon))
    )

    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
# YOUR CODE HERE
ax.hist(train_dist, bins=30, alpha=0.5, density=True, label='Training')
ax.hist(drift_dist, bins=30, alpha=0.5, density=True, label='Production')

ax.legend()
ax.set_title('Distribution Shift Visualization')
plt.tight_layout()
plt.show()
````

### Model Monitoring Concepts

- **Concept Drift:** The relationship between input features and the target variable changes over time, causing model performance to degrade.

- **Covariate Drift:** The distribution of input features changes over time, even if the relationship between inputs and outputs remains the same.

- **Population Stability Index (PSI) Thresholds:**
  - PSI < 0.10 → No significant shift (stable)
  - 0.10 ≤ PSI < 0.20 → Minor shift (monitor closely)
  - PSI ≥ 0.20 → Major shift (investigate and consider retraining)

- **Retraining Trigger:** Retrain the model when PSI indicates a major shift (≥ 0.20) or when production performance metrics fall below acceptable thresholds.

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = None
# YOUR CODE HERE
p_free = p_free_given_spam * p_spam + p_free_given_ham * p_ham

# Bayes
p_spam_given_free = None
# YOUR CODE HERE
p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # YOUR CODE HERE
    p_ham = 1 - prior_spam
    p_word = p_word_given_spam * prior_spam + p_word_given_ham * p_ham
    posterior = (p_word_given_spam * prior_spam) / p_word
    return posterior

pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

### Bayes' Theorem Terms in the Spam Example

- **Prior:** P(Spam) = 0.30  
  - The probability that an email is spam before seeing any words.

- **Likelihood:** P("free" | Spam) = 0.80  
  - The probability of observing the word "free" if the email is spam.

- **Evidence:** P("free")  
  - The overall probability of seeing the word "free" in any email.

- **Posterior:** P(Spam | "free")  
  - The probability that an email is spam after observing the word "free".

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30

sample_means = None
# YOUR CODE HERE — array of 5000 sample means
sample_means = np.array([
    np.mean(np.random.choice(population, size=sample_size))
    for _ in range(n_samples)
])

pop_mean = population.mean()
pop_std  = population.std()

clt_std  = None
# YOUR CODE HERE — predicted std of sample means
clt_std = pop_std / np.sqrt(sample_size)

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()

# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
ax.hist(sample_means, bins=30, density=True, alpha=0.6, label='Sample Means')

x = np.linspace(sample_means.min(), sample_means.max(), 300)
ax.plot(
    x,
    stats.norm.pdf(x, loc=pop_mean, scale=clt_std),
    linewidth=2,
    label='CLT Normal Approximation'
)

ax.legend()
ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout()
plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

The Central Limit Theorem (CLT) is important because it shows that the distribution of sample means becomes approximately normal as the sample size increases, even when the original population is not normally distributed. This allows statisticians and machine learning practitioners to make reliable inferences about population parameters using sample data.

The CLT justifies the normality assumption used in many classical statistical methods, including confidence intervals, hypothesis tests (such as z-tests and t-tests), and many regression-based inference procedures.
